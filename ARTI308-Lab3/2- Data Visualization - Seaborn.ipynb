# Seaborn Overview

Seaborn is a Python data visualization library based on matplotlib. It provides a high-level interface for drawing attractive and informative statistical graphics.




## Distribution Plots

Let's discuss some plots that allow us to visualize the distribution of a data set. These plots are:

* distplot
* jointplot
* pairplot
* rugplot
* kdeplot

## rugplot

rugplots are actually a very simple concept, they just draw a dash mark for every point on a univariate distribution. They are the building block of a KDE plot:

```python
sns.rugplot(tips['total_bill'])  # Create a rug plot showing all individual data points as marks on a single axis
```

## kdeplot

kdeplots are [Kernel Density Estimation plots](http://en.wikipedia.org/wiki/Kernel_density_estimation#Practical_estimation_of_the_bandwidth). These KDE plots replace every single observation with a Gaussian (Normal) distribution centered around that value. For example:

```python
# Code for visual demonstration of kernel density estimation concept
import numpy as np  # Import numpy for numerical operations
import matplotlib.pyplot as plt  # Import pyplot for low-level plotting
from scipy import stats  # Import stats for probability distributions

#Create dataset
dataset = np.random.randn(25)  # Generate 25 random values from standard normal distribution

# Create another rugplot
sns.rugplot(dataset);  # Create rug plot showing all 25 data points

# Set up the x-axis for the plot
x_min = dataset.min() - 2  # Set minimum x value: smallest data point minus 2
x_max = dataset.max() + 2  # Set maximum x value: largest data point plus 2

# 100 equally spaced points from x_min to x_max
x_axis = np.linspace(x_min,x_max,100)  # Create array of 100 evenly spaced points

# Set up the bandwidth, for info on this:
url = 'http://en.wikipedia.org/wiki/Kernel_density_estimation#Practical_estimation_of_the_bandwidth'  # Reference URL

bandwidth = ((4*dataset.std()**5)/(3*len(dataset)))**.2  # Calculate bandwidth using Scott's rule formula


# Create an empty kernel list
kernel_list = []  # Initialize empty list to store kernel functions

# Plot each basis function
for data_point in dataset:  # Loop through each of the 25 data points
    
    # Create a kernel for each point and append to list
    kernel = stats.norm(data_point,bandwidth).pdf(x_axis)  # Compute normal distribution pdf for this point
    kernel_list.append(kernel)  # Add this kernel to the list
    
    #Scale for plotting
    kernel = kernel / kernel.max()  # Normalize kernel by dividing by its maximum value
    kernel = kernel * .4  # Scale kernel to 40% of axis for visualization
    plt.plot(x_axis,kernel,color = 'grey',alpha=0.5)  # Plot the scaled kernel in light grey

plt.ylim(0,1)  # Set y-axis limits from 0 to 1
```

```python
# To get the kde plot we can sum these basis functions.

# Plot the sum of the basis function
sum_of_kde = np.sum(kernel_list,axis=0)  # Sum all kernels across the first axis to get combined density estimate

# Plot figure
fig = plt.plot(x_axis,sum_of_kde,color='indianred')  # Plot the summed kernel density in indianred color

# Add the initial rugplot
sns.rugplot(dataset,c = 'indianred')  # Overlay rug plot showing original data points

# Get rid of y-tick marks
plt.yticks([])  # Remove y-axis tick labels for cleaner appearance

# Set title
plt.suptitle("Sum of the Basis Functions")  # Add title to the figure
```

So with our tips dataset:

```python
sns.kdeplot(tips['total_bill'])  # Create kernel density estimation plot for total_bill column
sns.rugplot(tips['total_bill'])  # Overlay rug plot showing all individual data points
```

```python
sns.kdeplot(tips['tip'])  # Create kernel density estimation plot for tip column
sns.rugplot(tips['tip'])  # Overlay rug plot showing all individual tip data points
```

# Categorical Data Plots

Now let's discuss using seaborn to plot categorical data! There are a few main plot types for this:

* factorplot
* boxplot
* violinplot
* stripplot
* swarmplot
* barplot
* countplot

Let's go through examples of each!

```python
import seaborn as sns  # Import seaborn visualization library
%matplotlib inline  # Enable inline plotting in Jupyter notebooks
```

```python
tips = sns.load_dataset('tips')  # Load the tips dataset
tips.head()  # Display the first 5 rows of the dataset
```

## barplot and countplot

These very similar plots allow you to get aggregate data off a categorical feature in your data. **barplot** is a general plot that allows you to aggregate the categorical data based off some function, by default the mean:

```python
sns.barplot(x='sex',y='total_bill',data=tips)  # Create bar plot showing average total_bill for each sex category
```

```python
import numpy as np  # Import numpy for using standard deviation estimator
```

You can change the estimator object to your own function, that converts a vector to a scalar:

```python
sns.barplot(x='sex',y='total_bill',data=tips,estimator=np.std)  # Create bar plot showing standard deviation of total_bill for each sex
```

### countplot

This is essentially the same as barplot except the estimator is explicitly counting the number of occurrences. Which is why we only pass the x value:

```python
sns.countplot(x='sex',data=tips)  # Create count plot showing number of observations for each sex category
```

## boxplot and violinplot

boxplots and violinplots are used to shown the distribution of categorical data. A box plot (or box-and-whisker plot) shows the distribution of quantitative data in a way that facilitates comparisons between variables or across levels of a categorical variable. The box shows the quartiles of the dataset while the whiskers extend to show the rest of the distribution, except for points that are determined to be “outliers” using a method that is a function of the inter-quartile range.

```python
sns.boxplot(x="day", y="total_bill", data=tips,palette='rainbow')  # Create box plot of total_bill by day with rainbow colored palette
```

```python
# Create box plot for entire dataframe with horizontal orientation
sns.boxplot(data=tips,palette='rainbow',orient='h')  # Create horizontal box plots for all numerical columns in tips data
```

```python
sns.boxplot(x="day", y="total_bill", hue="smoker",data=tips, palette="coolwarm")  # Create box plot split by smoker status with coolwarm palette
```

### violinplot
A violin plot plays a similar role as a box and whisker plot. It shows the distribution of quantitative data across several levels of one (or more) categorical variables such that those distributions can be compared. Unlike a box plot, in which all of the plot components correspond to actual datapoints, the violin plot features a kernel density estimation of the underlying distribution.

```python
sns.violinplot(x="day", y="total_bill", data=tips,palette='rainbow')  # Create violin plot of total_bill by day using rainbow palette
```

```python
sns.violinplot(x="day", y="total_bill", data=tips,hue='sex',palette='Set1')  # Create violin plot split by sex using Set1 palette
```

```python
sns.violinplot(x="day", y="total_bill", data=tips,hue='sex',split=True,palette='Set1')  # Create split violin plot: male and female violins side-by-side
```

## stripplot and swarmplot
The stripplot will draw a scatterplot where one variable is categorical. A strip plot can be drawn on its own, but it is also a good complement to a box or violin plot in cases where you want to show all observations along with some representation of the underlying distribution.

The swarmplot is similar to stripplot(), but the points are adjusted (only along the categorical axis) so that they don’t overlap. This gives a better representation of the distribution of values, although it does not scale as well to large numbers of observations (both in terms of the ability to show all the points and in terms of the computation needed to arrange them).

```python
sns.stripplot(x="day", y="total_bill", data=tips, palette='rainbow')  # Create strip plot (scatter) of total_bill by day with rainbow colors
```

```python
sns.stripplot(x="day", y="total_bill", data=tips,jitter=True, palette='rainbow')  # Create strip plot with jitter to prevent point overlap
```

```python
sns.stripplot(x="day", y="total_bill", data=tips,jitter=True,hue='sex',palette='Set1')  # Create strip plot with jitter and color split by sex
```

```python
sns.stripplot(x="day", y="total_bill", data=tips,jitter=True,hue='sex',palette='Set1',dodge=True)  # Create strip plot with male and female points separated
```

```python
sns.swarmplot(x="day", y="total_bill", data=tips)  # Create swarm plot with non-overlapping points arranged to show distribution
```

### Combining Categorical Plots

```python
sns.violinplot(x="tip", y="day", data=tips,palette='rainbow')  # Create violin plot showing tip distribution by day
sns.swarmplot(x="tip", y="day", data=tips,color='black',size=3)  # Overlay swarm plot with black dots showing individual observations
```

## catplot

factorplot is the most general form of a categorical plot. It can take in a kind parameter to adjust the plot type:

```python
sns.catplot(x='sex',y='total_bill',data=tips,kind='bar')  # Create categorical plot (general form) showing bar plot of total_bill by sex
```

Matrix plots allow you to plot data as color-encoded matrices and can also be used to indicate clusters within the data (later in the machine learning section we will learn how to formally cluster data).

Let's begin by exploring seaborn's heatmap and clutermap:

```python
import seaborn as sns  # Import seaborn visualization library
%matplotlib inline  # Enable inline plotting for displaying plots in notebooks
```

## Heatmap

In order for a heatmap to work properly, your data should already be in a matrix form, the sns.heatmap function basically just colors it in for you. For example:

```python
tips.head()  # Display first 5 rows of tips data to review before creating heatmap
```

```python
# Display correlation matrix for all numerical columns
tips.corr()  # Calculate Pearson correlation coefficients between all numerical variables
```

```python
sns.heatmap(tips.corr())  # Create a heatmap showing the correlation matrix of tips dataset
```

```python
sns.heatmap(tips.corr(),cmap='coolwarm',annot=True)  # Create heatmap with coolwarm colormap and annotations showing correlation values
```

Or for the flights data:

```python
flights.pivot_table(values='passengers',index='month',columns='year')  # Create pivot table with months as rows, years as columns, passenger counts as values
```

```python
pvflights = flights.pivot_table(values='passengers',index='month',columns='year')  # Create pivot table and store it
sns.heatmap(pvflights)  # Create heatmap from the pivot table showing passenger trends
```

```python
sns.heatmap(pvflights,cmap='magma',linecolor='white',linewidths=1)  # Create heatmap with magma colormap and white gridlines separating cells
```

## clustermap

The clustermap uses hierarchal clustering to produce a clustered version of the heatmap. For example:

```python
sns.clustermap(pvflights)  # Create clustered heatmap with hierarchical clustering to group similar months and years
```

Notice now how the years and months are no longer in order, instead they are grouped by similarity in value (passenger count). That means we can begin to infer things from this plot, such as August and July being similar (makes sense, since they are both summer travel months)

```python
# Create clustered heatmap with normalization to highlight patterns more clearly
sns.clustermap(pvflights,standard_scale=1)  # Create clustermap with standardized data (z-score normalization) for columns
```

# Regression Plots

Seaborn has many built-in capabilities for regression plots, however we won't really discuss regression until the machine learning section of the course, so we will only cover the **lmplot()** function for now.

**lmplot** allows you to display linear models, but it also conveniently allows you to split up those plots based off of features, as well as coloring the hue based off of features.

Let's explore how this works:

```python
import seaborn as sns  # Import seaborn for regression plotting
%matplotlib inline  # Enable inline plotting in notebooks
```

```python
tips = sns.load_dataset('tips')  # Load the tips dataset
```

```python
tips.head()  # Display first 5 rows of the tips dataset
```

## lmplot()

```python
sns.lmplot(x='total_bill',y='tip',data=tips)  # Create linear model plot showing regression line between total_bill and tip
```

```python
sns.lmplot(x='total_bill',y='tip',data=tips,hue='sex')  # Create lmplot with separate regression lines for male and female
```

```python
sns.lmplot(x='total_bill',y='tip',data=tips,hue='sex',palette='coolwarm')  # Create lmplot with separate lines colored by sex using coolwarm palette
```

## Using a Grid

We can add more variable separation through columns and rows with the use of a grid. Just indicate this with the col or row arguments:

```python
sns.lmplot(x='total_bill',y='tip',data=tips,col='sex')  # Create side-by-side lmplots in columns separated by sex variable
```

```python
sns.lmplot(x="total_bill", y="tip", row="sex", col="time",data=tips)  # Create grid of lmplots with rows for sex and columns for time
```

```python
sns.lmplot(x='total_bill',y='tip',data=tips,col='day',hue='sex',palette='coolwarm')  # Create lmplot with columns for each day and hue for sex
```

## Aspect and Size

Seaborn figures can have their size and aspect ratio adjusted with the **height** and **aspect** parameters:

```python
sns.lmplot(x='total_bill',y='tip',data=tips,col='day',hue='sex',palette='coolwarm',
          aspect=0.6,height=8)  # Create lmplot with custom aspect ratio (narrower) and height (taller)
```
